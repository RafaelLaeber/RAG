```python
from agentic_rag.retrieval.vectorstore import get_vector_store

vector_store = get_vector_store()
```

```python
question = "Qual o limite de reembolso para refeição em viagem?"

results = vector_store.similarity_search_with_score(question, k=4)

for doc, score in results:
    print(f"{score:.4f} | {doc.metadata['source']} | página {doc.metadata.get('page')}")
    print(f"   {doc.page_content[:120]!r}")
    print()
```

```python
results = vector_store.similarity_search_with_score(
    question,
    k=4,
    filter={"source": "orbita_politica_reembolso_2026.md"},
)

for doc, score in results:
    print(f"{score:.4f} | {doc.metadata['source']}")
    print(
        f"   {doc.page_content[:120]!r}"
    )  # Como está só 120 caracteres não apareceu a informação que queriamos
    print()
```
