```python
def cosine_similarity(a: list[float], b: list[float]) -> float:
    """Calcula a similaridade de cosseno entre dois vetores."""
    a_arr = np.array(a)
    b_arr = np.array(b)
    return float(a_arr @ b_arr / (np.linalg.norm(a_arr) * np.linalg.norm(b_arr)))


print(np.linalg.norm(vector))
```

```python
sentences = [
    "O gato dormiu no sofá.",
    "O felino descansou no estofado.",
    "The cat slept on the couch.",
    "A bolsa de valores caiu hoje.",
    "Eu adoro programar em Python.",
]

vectors = embeddings.embed_documents(sentences)
reference = vectors[0]

for sentence, vec in zip(sentences, vectors, strict=True):
    print(f"{cosine_similarity(reference, vec):.3f}  {sentence}")
```

```python
knowledge_base = [
    "O Astra DB é um banco de dados vetorial gerenciado na nuvem.",
    "LCEL usa o operador pipe para compor Runnables.",
    "Embeddings transformam texto em vetores numéricos.",
    "O arroz deve ser lavado antes do cozimento.",
    "Python é uma linguagem de programação muito usada em IA.",
]
kb_vectors = embeddings.embed_documents(knowledge_base)


def search(query: str, k: int = 2) -> list[tuple[float, str]]:
    """Retorna os k textos mais parecidos com a pergunta."""
    query_vector = embeddings.embed_query(query)
    scores = [cosine_similarity(query_vector, vec) for vec in kb_vectors]
    ranked = sorted(zip(scores, knowledge_base, strict=True), reverse=True)
    return ranked[:k]


search("Onde eu guardo meus vetores?")
```
